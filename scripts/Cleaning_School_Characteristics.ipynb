```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm
import warnings
warnings.filterwarnings("ignore")
```

```python
dat_satact = pd.read_csv("SAT and ACT Cleaned.csv")
char = pd.read_csv("School Characteristics.csv")
```

```python
keys = dat_satact['Combokey']
```

```python
char = char.loc[char['COMBOKEY'].isin(keys)]
```

```python
char['SCH_MAGNETDETAIL'].unique()
```

```python
# formatting NA values
char = char.replace('-9', np.nan,)
char = char.replace('-10', np.nan)
char = char.replace(-10, np.nan)


char = char.dropna(axis = 1, how = 'all')
char
```

```python
char = char.drop(columns = ['LEA_STATE', 'LEA_STATE_NAME', 'LEAID', 'LEA_NAME', 'SCHID', 'SCH_NAME',
                    'SCH_JUST', 'SCH_JUST_IND', 'SCH_GRADE_PS',
       'SCH_GRADE_KG', 'SCH_GRADE_G01', 'SCH_GRADE_G02', 'SCH_GRADE_G03',
       'SCH_GRADE_G04', 'SCH_GRADE_G05', 'SCH_GRADE_G06', 'SCH_GRADE_G07',
       'SCH_GRADE_G08', 'SCH_GRADE_G09', 'SCH_GRADE_G10', 'SCH_GRADE_G11',
       'SCH_GRADE_G12','SCH_MAGNETDETAIL', 'SCH_ALTFOCUS'])
```

```python
char = char.rename(columns={'COMBOKEY': 'Combokey',
                     'SCH_VIRT_IND': 'Fully Virtual',
                     'SCH_GRADE_UG': 'Ungraded',
                     'SCH_STATUS_SPED': 'Special Education',
                     'SCH_STATUS_MAGNET': 'Magnet School/Program',
                     'SCH_STATUS_CHARTER': 'Charter School',
                     'SCH_STATUS_ALT': 'Alternative School'})
```
